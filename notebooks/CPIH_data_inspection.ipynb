### CPIH Historic Inflation Data
Downloads as an Excel from:
https://www.ons.gov.uk/file?uri=/economy/inflationandpriceindices/datasets/consumerpriceinflation/current/consumerpriceinflationdetailedreferencetables.xlsx

Relevant time series is Table 37.

Uses CPIH since RPI is no longer considered for inflation by ONS.


```python
import pandas as pd
```

```python
path = '../data/raw/consumerpriceinflationdetailedreferencetables.xlsx'

df = pd.read_excel(path, sheet_name = 'Table 37',skiprows=6)
df.head()
```

#### Cleaned output

```python
df = df[['name','CPIH ALL ITEMS']]
df.head()
```

```python
df.to_csv('../data/cpih_clean.csv')
```
